# Errors

## Question

Three students measure the same resistance several times, with final measurements (in Ohms):

Student 1: $R = 11 \pm 1$
Student 2: $R = 12 \pm 1$
Student 3: $R = 10 \pm 3$.

Given these three results what do we write down as the best estimate for the resistance $R$? Tip: you will need to use weighted errors.

### Answer

*Your answer here.*

```python

def calculate_weights(sigma):
  weights = []
  for x in sigma:
    weights.append(1/(x**2))
  #print(weights)
  return weights

def calculate_sum_of_weights_times_measurement(measurements,weights):
  sum = 0
  for x in range(0,len(measurements)):
    sum = sum + (weights[x] * measurements[x])

  return sum

def calculate_sum_of_weights(weights):
  sum = 0
  for x in weights:
    sum = sum + x
  return sum

def calculate_estimate(sigmas, measurements):
  weights = calculate_weights(sigmas)
  num = calculate_sum_of_weights_times_measurement(measurements,weights)
  den = calculate_sum_of_weights(weights)
  #print(num)
  #print(den)

  return num/den

sigmas = [1,1,3]
measurements = [11,12,10]
print(calculate_estimate(sigmas,measurements))
```
