# Chi-square test

## Question

You publish flyers for an event in three different colours and want to test whether the colour has a significant affect on whether people take them. The data is available below.  Use the $\chi^2$ test to do this, and calculate the probability of the significance of the result.

| |white | light blue | neon pink | Totals|
|---|---|---|---|---|
|Take | 32| 38 | 20| 90|
|Don't take | 8| 22 |30 | 60|
|Totals | 40|60 | 50| 150|

### Answer

```python
#values from table
#take, don't take, expected take, expected dont take
white = [32,8,24,16]
blue = [38,22,36,24]
pink = [20,30,30,20]
test = [white,blue,pink]

#calculate chi
def calculate_chi(observed,expected):
  chi = ((observed - expected)**2) / expected
  return chi

chi = 0
for x in test:
  take = calculate_chi(x[0],x[2])
  dont_take = calculate_chi(x[1],x[3])
  chi = chi + take + dont_take

print('chi is {:.3f}'.format(chi))

#degrees of freedom
#(rows - 1) * (column - 1) = (2-1) * (2-1) = 1
dof = 1

print("Probability is between 0.001 and 0.0001")
```
